```python
import pickle
loaded_model=pickle.load(open('finalckd_model.sav','rb'))
age=float(input("Enter Age : "))
bp=float(input("Enter bp :"))
al=float(input("Enter al :"))
su=float(input("Enter su :"))
bgr=float(input("Enter bgr :")) 
bu=float(input("Enter bu :"))
sc=float(input("Enter sc :")) 
sod=float(input("Enter sod :"))
pot=float(input("Enter pot :"))
hrmo=float(input("Enter hrmo :")) 
pcv=float(input("Enter pcv :")) 
wc=float(input("Enter wc :")) 
rc=float(input("Enter rc :")) 
sg_b=int(input("Enter 1 if sg is b; otherwise 0 :"))
sg_c=int(input("Enter 1 if sg is c; otherwise 0 :"))
sg_d=int(input("Enter 1 if sg is d; otherwise 0 :"))
sg_e=int(input("Enter 1 if sg is e; otherwise 0 :"))
rbc_normal=int(input("Enter rbc 1 if normal; otherwise 0 :"))
pc_normal=int(input("Enter pc 1 if normal; otherwise 0 :"))
pcc_present=int(input("Enter pcc 1 if present; otherwise 0 :"))
ba_present=int(input("Enter ba 1 if present; otherwise 0 :"))
htn_yes=int(input("Enter htn 1 if yes; otherwise 0 :"))
dm_yes=int(input("Enter dm 1 if yes; otherwise 0 :"))
cad_yes=int(input("Enter cad 1 if yes; otherwise 0 :"))
appet_yes=int(input("Enter appet 1 if yes; otherwise 0 :"))
pe_yes=int(input("Enter pe 1 if yes; otherwise 0 :"))
ane_yes=int(input("Enter ane 1 if yes; otherwise 0 :"))
result=loaded_model.predict([[age,bp,al,su,bgr,bu,sc,sod,pot,hrmo,pcv,wc,rc,sg_b,sg_c,sg_d,sg_e,rbc_normal,pc_normal,pcc_present,ba_present,htn_yes,dm_yes,cad_yes,appet_yes,pe_yes,ane_yes]])
if(result==1):
    print("This person has been diagnoised with Chronic Kidney Disease")
else:
    print("This person  has not been diagnoised with Chronic Kidney Disease")
```
