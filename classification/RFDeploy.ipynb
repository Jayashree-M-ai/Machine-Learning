```python
import pickle
loaded_model=pickle.load(open('final_model.sav','rb'))
age=int(input("Enter Age : "))
sal=float(input("Enter Estimated_Salary : "))
gen=bool(input("Enter 1 for male; 0 for female : "))
result=loaded_model.predict([[age,sal,gen]])
if(result==1):
    print("This person will purchase")
else:
    print("This person will not purchase")
```
