```python
#Data collection
import pandas as pd
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
#Data preprocession to change categorial column-nominal value into one hot encoding
dataset=pd.get_dummies(dataset, dtype=int, drop_first=True)
dataset
```

```python
#input split
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
independent
```

```python
#output split
dependent=dataset[['charges']]
dependent
```

```python
#train and test set
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.30,random_state=0)
```

```python
#model creation
import xgboost as xgb
regressor= xgb.XGBRegressor(objective='reg:squarederror', n_estimators=100, learning_rate=0.1, random_state=0)
regressor.fit(x_train,y_train)
```

```python
#evaluation
y_pred=regressor.predict(x_test)
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
r_score
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 6))
sns.regplot(x=y_test, y=y_pred, scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("XGBoost Regressor: Actual vs. Predicted Values for insurance charges prediction")
plt.grid(True)
plt.show()
```
