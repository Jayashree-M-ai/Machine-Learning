```python
#Data collection
import pandas as pd
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
#Data preprocession to change categorial column-nominal value into one hot encoding
dataset=pd.get_dummies(dataset, dtype=int, drop_first=True)
dataset
```

```python
#input split
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
independent
```

```python
#output split
dependent=dataset[['charges']]
dependent
```

```python
#model creation
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
param_grid={ 'kernel' : ['rbf','linear'] ,'C':[10,100,1000,3000], 'gamma':['auto','scale']}
grid=GridSearchCV(SVR(),param_grid,refit=True,verbose=1,n_jobs=2)
grid.fit(independent,dependent)
```

```python
res=grid.cv_results_
print("The R_Score value for best parameter{}".format(grid.best_params_))
```

```python
table=pd.DataFrame.from_dict(res)
table
```

```python
age=float(input("Enter Age :"))
bmi=float(input("Enter BMI :"))
child=int(input("Enter number of children :"))
sex=int(input("Enter 1 for male and 0 for female :"))
smoker=int(input("Enter 1 for smoker and 0 for non-smoker :"))
Insurance_charges_predicted=grid.predict([[age,bmi,child,sex,smoker]])
print("Insurance_charges_predicted={}".format(Insurance_charges_predicted))
```
