```python
import pickle
```

```python
sc=pickle.load(open("preprocessing.sav",'rb'))
scy=pickle.load(open("preprocessingy.sav",'rb'))
```

```python
R_D=float(input("Enter R & D investment"))
Admin=float(input("Enter Administration investment"))
Mar=float(input("Enter Marketing investment"))
S_F=bool(input("Enter 1 if invested in Florida otherwise enter 0"))
S_N=bool(input("Enter 1 if invested in Newyork otherwise enter 0"))
preinput=sc.transform([[R_D,Admin,Mar,S_F,S_N]])
loaded_model=pickle.load(open("final_model.sav",'rb'))
result=loaded_model.predict(preinput)
preoutput=scy.inverse_transform([result])
print("Profit Prediction=",float(preoutput))
```
