```python
import pickle
import skfuzzy as fuzz
import numpy as np
loaded_model=pickle.load(open('final_model.sav','rb'))
ai=float(input("Enter Annual income : "))
ss=int(input("Enter Spending Score : "))
inputs=np.array([[ai,ss]])
m = 1.5
error = 0.005
maxiter = 1000
u, u0, d, jm, p, fpc = fuzz.cmeans_predict(inputs.T,loaded_model, m=m, error=error, maxiter=maxiter, init=None)
print("membership values for new input:",u)
labels = np.argmax(u, axis=0)
print("Predicted cluster:",labels)
```
